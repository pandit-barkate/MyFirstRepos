```python
import pandas as pd

import seaborn as sns # to get the dataset for multi class classification
```

```python
data = sns.load_dataset('iris')
```

```python
data
```

```python
data['species'].value_counts()
```

```python
# divide data in X and y
X = data.drop(['species'],axis=1)
```

```python
X
```

```python
y = data[['species']]
```

```python
y
```

```python
# initialize one hot encoder for miltucalss equal number of class
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder(sparse_output=False) # values in the diagonal is 1 in matrix
```

```python
y_transform = encoder.fit_transform(y)  # expect two dimesion so add [[]] for y
```

```python
y_transform
```

```python
# scaling feature in DL
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_transform = scaler.fit_transform(X)
```

```python
# Define ANN arcitecture
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense # for mlp alsways desn

# initialize model
model = Sequential()

# 1st layer => i/p layer
model.add(Dense(units=4,activation='relu')) # nuron = number of feature

#2nd hiden layer
model.add(Dense(units=8,activation='relu'))  # souble of first

#3rd layer
model.add(Dense(units=16,activation='relu'))

# final layer - Op/p layer
model.add(Dense(units=3,activation='softmax'))  # nuron = number of classes
```

```python
model.summary()
```

```python
# model.compile

model.compile(optimizer='adam',loss='categorical_crossentropy',metrics=['accuracy'])
# loss for multi class categorical_crossentropy
# loss for regression is mse
```

```python
# split data in train test
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X_transform,y_transform,test_size=0.2)
```

```python
# training model
model.fit(X_train,y_train,epochs=100)  # model is trained 100 times on same data
```

```python
#  model Eval
import numpy as np
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
# above y_pred is the probability of all 3 classes
# convert these probability to valeus
import numpy as np
y_pred_Argmax = np.argmax(y_pred,axis=1)

# argmax() finds higest value and return the index value and the valur itself
```

```python
y_pred_Argmax
```

```python
y_test
```

```python
# so convert this y_test to argmax

y_test_argmax = np.argmax(y_test,axis=1)
```

```python
y_test_argmax
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test_argmax,y_pred_Argmax))
```
